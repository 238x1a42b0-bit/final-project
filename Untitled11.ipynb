```python

from pypdf import PdfReader

pdf_path = "/content/Crop_Disease_RAG_Knowledge_Base (1) (3).pdf"

reader = PdfReader(pdf_path)

pages = []

for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""
    pages.append({
        "page": page_number,
        "text": text
    })

print("Total pages:", len(pages))
print("\nFirst page text:\n")
print(pages[0]["text"][:2000])
```

```python

# Step 3: Split PDF text into smaller chunks

CHUNK_SIZE = 1200
CHUNK_OVERLAP = 200

chunks = []

for page in pages:
    text = page["text"]
    start = 0

    while start < len(text):
        end = start + CHUNK_SIZE
        chunk_text = text[start:end].strip()

        if chunk_text:
            chunks.append({
                "text": chunk_text,
                "page": page["page"]
            })

        start += CHUNK_SIZE - CHUNK_OVERLAP

print("Total pages:", len(pages))
print("Total chunks created:", len(chunks))

# Display the first chunk
print("\nFirst chunk:\n")
print(chunks[0]["text"])
```

```python

from sentence_transformers import SentenceTransformer

# Load the embedding model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# Extract text from chunks
chunk_texts = [chunk["text"] for chunk in chunks]

# Convert text into numerical vectors
embeddings = model.encode(
    chunk_texts,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("Total chunks:", len(chunk_texts))
print("Embeddings shape:", embeddings.shape)
print("\nFirst embedding:")
print(embeddings[0][:10])
```

```python

import chromadb

print("ChromaDB installed successfully!")
print("Version:", chromadb.__version__)
```

```python

client = chromadb.Client()

collection = client.create_collection(
    name="crop_disease_knowledge"
)

ids = [f"chunk_{i}" for i in range(len(chunks))]
documents = [chunk["text"] for chunk in chunks]

metadatas = [
    {"page": int(chunk["page"])}
    for chunk in chunks
]

collection.add(
    ids=ids,
    documents=documents,
    embeddings=embeddings.tolist(),
    metadatas=metadatas
)

print("Vector database created successfully!")
print("Total stored chunks:", collection.count())
```

```python

# Step 6: Retrieve relevant information

question = "What are the symptoms of tomato early blight?"

# Convert the question into an embedding
query_embedding = model.encode(
    [question],
    normalize_embeddings=True
).tolist()

# Search the vector database
results = collection.query(
    query_embeddings=query_embedding,
    n_results=3
)

print("Question:", question)
print("\nRetrieved information:\n")

for i, document in enumerate(
    results["documents"][0], start=1
):
    print(f"Result {i}:")
    print(document[:500])
    print("-" * 50)
```

```python

from groq import Groq
from getpass import getpass

api_key = getpass("Enter your Groq API key: ")

client_llm = Groq(api_key=api_key)

print("LLM client connected successfully!")
```

```python

# Step 7C: Generate an answer using RAG

question = "What are the symptoms of tomato early blight?"

# 1. Convert question into an embedding
query_embedding = model.encode(
    [question],
    normalize_embeddings=True
).tolist()

# 2. Retrieve relevant information from ChromaDB
results = collection.query(
    query_embeddings=query_embedding,
    n_results=3
)

# 3. Combine retrieved information
context = "\n\n".join(results["documents"][0])

# 4. Generate an answer using the LLM
response = client_llm.chat.completions.create(
    model="llama3-70b-8192",
    messages=[
        {
            "role": "system",
            "content": (
                "You are a crop disease assistant. "
                "Answer using only the supplied context. "
                "If the context does not contain enough "
                "information, clearly say so."
            )
        },
        {
            "role": "user",
            "content": (
                f"Context:\n{context}\n\n"
                f"Question: {question}"
            )
        }
    ],
    temperature=0.2
)

# 5. Display the final answer
print("Question:", question)
print("\nRAG Answer:\n")
print(response.choices[0].message.content)
```

```python

def ask_crop_question(question):
    # 1. Convert question into embedding
    query_embedding = model.encode(
        [question],
        normalize_embeddings=True
    ).tolist()

    # 2. Retrieve relevant information
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=3
    )

    # 3. Prepare context
    context = "\n\n".join(results["documents"][0])

    # 4. Generate answer using LLM
    response = client_llm.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a crop disease assistant. "
                    "Answer using the provided context only. "
                    "If information is missing, say so. "
                    "Give clear, simple explanations."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Context:\n{context}\n\n"
                    f"Question: {question}"
                )
            }
        ],
        temperature=0.2
    )

    print("Question:", question)
    print("\nAnswer:\n")
    print(response.choices[0].message.content)


# Test your chatbot
ask_crop_question("How can I identify potato late blight?")
```

```python

# Interactive Crop Disease RAG Chatbot

while True:
    question = input(
        "\nAsk a crop disease question "
        "(type 'exit' to stop): "
    )

    if question.lower().strip() == "exit":
        print("Thank you for using Crop Disease Assistant!")
        break

    ask_crop_question(question)
```
